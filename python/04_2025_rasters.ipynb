# 2025 Dynamic Covariates

Extracts dynamic covariates from 2025 for a 2025 prediction surface. 

All layers exported at 1km resolution, EPSG:4326, to Google Drive (`sudan_enm_covariates`).

```python
import ee
import geemap

ee.Initialize(project="sudan-enm")

study_extent = ee.Geometry.Rectangle([21.5, 8.5, 39, 22.5])

export_params = {
    'folder': 'sudan_enm_covariates',
    'region': study_extent,
    'scale': 1000,
    'crs': 'EPSG:4326',
    'maxPixels': 1e9
}
```

```python
# Night LST
def scale_lst(image):
    return image.multiply(0.02).subtract(273.15) \
        .copyProperties(image, ['system:time_start'])

lst_night_2025 = ee.ImageCollection('MODIS/061/MOD11A2') \
    .select('LST_Night_1km') \
    .filterBounds(study_extent) \
    .filterDate('2025-01-01', '2026-01-01') \
    .map(scale_lst) \
    .mean() \
    .clip(study_extent)

ee.batch.Export.image.toDrive(
    image=lst_night_2025,
    description='lst_night_annual_2025_1km',
    **export_params
).start()

print("LST night 2025 export started")
```

```python
# Rainfall
rainfall_2025 = ee.ImageCollection('UCSB-CHG/CHIRPS/DAILY') \
    .filterBounds(study_extent) \
    .filterDate('2025-01-01', '2026-01-01') \
    .sum() \
    .clip(study_extent)

ee.batch.Export.image.toDrive(
    image=rainfall_2025,
    description='rainfall_2025_1km',
    **export_params
).start()

print("Rainfall 2025 export started")
```
